```python
import re
import unicodedata
from langdetect import detect, LangDetectException
```

```python
# =====================================================================
# COMPILE REGEX TOÀN CỤC CHO BỘ LỌC UNICODE NGOẠI NGỮ (TỐI ƯU TỐC ĐỘ)
# =====================================================================
UNICODE_PATTERNS = [
    r'[\u4e00-\u9fff]',       # Tiếng Trung
    r'[\u3040-\u30ff]',       # Tiếng Nhật
    r'[\uac00-\ud7af]',       # Tiếng Hàn
    r'[\u0600-\u06FF]',       # Tiếng Ả Rập
    r'[\u0e00-\u0e7f]',       # Tiếng Thái
    r'[\u1000-\u109f]',       # Tiếng Myanmar
    r'[\u0900-\u097f]',       # Tiếng Ấn Độ (Devanagari)
    r'[\u0b80-\u0bff]',       # Tiếng Ấn Độ (Tamil)
    r'[\u0400-\u04FF]',       # Tiếng Nga
    # Tiếng Việt có dấu NFC (VADER không hiểu nên cần lọc bỏ)
    r'[àáảãạăằắẳẵặâầấẩẫậèéẻẽẹêềếểễệđìíỉĩịòóỏõọôồốổỗộơờớởỡợùúủũụưừứửữựỳýỷỹỵ]'
]
# Gom nhóm các dải ngôn ngữ và biên dịch 1 lần duy nhất
FOREIGN_LANG_PATTERN = re.compile('|'.join(UNICODE_PATTERNS), flags=re.IGNORECASE)


def is_foreign_unicode(text: str) -> bool:
    """Kiểm tra nhanh xem văn bản có chứa ký tự của các ngôn ngữ phi-Anh không."""
    # CHỐT CHẶN 1: Nếu dòng bị lỗi kiểu dữ liệu (không phải str), báo False để xử lý sau
    if not isinstance(text, str):
        return False

    # Chuẩn hóa NFC trước khi check để tránh bỏ sót chữ tiếng Việt tổ hợp dạng NFD
    norm_text = unicodedata.normalize('NFC', text)
    return bool(FOREIGN_LANG_PATTERN.search(norm_text.lower()))


def is_english_langdetect(text: str) -> bool:
    """Sử dụng mô hình langdetect để kiểm tra xem văn bản có phải tiếng Anh không."""
    # CHỐT CHẶN 2: Kiểm tra kiểu dữ liệu an toàn
    if not isinstance(text, str):
        return False

    # Loại bỏ các ký tự phi ngôn ngữ để mô hình langdetect nhận diện chuẩn xác nhất
    clean_for_detect = re.sub(r'[^a-zA-Z\s]', '', text).strip()

    # TỐI ƯU: Nếu chuỗi quá ngắn sau khi lọc ký tự đặc biệt, mặc định giữ lại
    # vì langdetect thường đoán sai tỷ lệ cao trên chuỗi ngắn (ví dụ "BTC moon" -> đoán bừa)
    if len(clean_for_detect) < 5:
        return True

    try:
        return detect(clean_for_detect) == 'en'
    except LangDetectException:
        # Nếu xảy ra lỗi nhận diện (chuỗi rác), trả về False để loại bỏ
        return False


def language_filter(df: pd.DataFrame, report: bool = True) -> pd.DataFrame:
    """
    Bước 5 — Lọc ngôn ngữ phi Anh (Hybrid Language Filter - Bản sửa lỗi TypeError).
    """
    df = df.copy()
    initial_count = len(df)

    # =====================================================================
    # [GIA CỐ BẢO VỆ] Loại bỏ các dòng bị NaN hoặc không phải kiểu chuỗi (str)
    # =====================================================================
    df = df.dropna(subset=['text_clean'])
    df = df[df['text_clean'].apply(lambda x: isinstance(x, str))]
    df = df[df['text_clean'].str.strip().str.len() > 0]
    dropped_nan = initial_count - len(df)

    # -----------------------------------------------------------------
    # MÀNG LỌC 1: Regex Unicode (Chạy cực kỳ nhanh bằng C-level regex)
    # -----------------------------------------------------------------
    before_unicode = len(df)
    is_foreign = df['text_clean'].apply(is_foreign_unicode)
    df = df[~is_foreign]
    dropped_unicode = before_unicode - len(df)

    # -----------------------------------------------------------------
    # MÀNG LỌC 2: Langdetect (Chạy chậm hơn nên chỉ quét trên tệp dữ liệu đã lọc sạch ở Màng 1)
    # -----------------------------------------------------------------
    before_langdetect = len(df)
    is_english = df['text_clean'].apply(is_english_langdetect)
    df = df[is_english]
    dropped_langdetect = before_langdetect - len(df)

    final_count = len(df)
    total_removed = initial_count - final_count
    retention = (final_count / initial_count) * 100 if initial_count > 0 else 0

    if report:
        print('=' * 65)
        print('    BÁO CÁO LỌC NGÔN NGỮ (HYBRID LANGUAGE FILTER) — BƯỚC 5')
        print('=' * 65)
        print(f"  Tweet đầu vào                     : {initial_count:>10,}")
        print(f"  Loại bỏ do NaN / Rỗng ẩn          : {dropped_nan:>10,}")
        print(f"  Loại bằng Regex Unicode (Màng 1)  : {dropped_unicode:>10,}")
        print(f"  Loại bằng Langdetect (Màng 2)     : {dropped_langdetect:>10,}")
        print(f"  Tổng số tweet ngoại ngữ bị loại    : {total_removed:>10,}")
        print('-' * 65)
        print(f"  Tweet tiếng Anh hợp lệ giữ lại    : {final_count:>10,}")
        print(f"  Tỉ lệ giữ lại của bước này        : {retention:>9.1f}%")
        print('=' * 65)

    return df.reset_index(drop=True)
```
